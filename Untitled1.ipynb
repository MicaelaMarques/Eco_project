```python
data = customers['Improvement_Areas']
def categorize_area(area):
    target_groups = [
        'Customer service responsiveness',
        'Product quality and accuracy',
        'Reducing packaging waste',
        'Shipping speed and reliability'
    ]
    if area in target_groups:
        return area
    else:
        return 'Others'
customers['Improvement_Areas_New'] = customers['Improvement_Areas'].apply(categorize_area)
customers
```

```python
#customers_reduce_waste_only = [customer[
customers['Improvement_Areas_New'] == 'Reducing packaging waste'
```

```python
improvement_areas_grouped = customers.groupby(['category_split', 'Improvement_Areas_New'], as_index=False).agg({'customer_id':'sum'})
#display(improvement_areas_grouped)
#improvement_areas_grouped
improvement_areas_grouped = improvement_areas_grouped[improvement_areas_grouped['Improvement_Areas_New'] == 'Reducing packaging waste']
#improvement_areas_grouped['Improvement_Areas_New']
improvement_areas_grouped = improvement_areas_grouped.rename(columns={'customer_id':'customer_count'})
```

```python
improvement_areas_grouped
```

```python
amazon_improvement_areas = pd.merge(amazon, improvement_areas_grouped, right_on='category_split', left_on='general_category', how='left')
```

```python
amazon_imp = amazon_improvement_areas.groupby(['green', 'Improvement_Areas_New'], as_index=False)['customer_count'].mean
```
