```python
import plotly.express as px
```

```python
import pandas as pd

customer_behavior = pd.read_csv('/content/Amazon Customer Behavior Survey.csv')
```

```python
import uuid

# Generate a UUID for each customer
customer_behavior['customer_id'] = [str(uuid.uuid4()) for _ in range(len(customer_behavior))]
```

```python
customer_behavior['Purchase_Categories'].unique()
```

```python
customer_behavior['category_split'] = customer_behavior['Purchase_Categories'].str.split(';')

# Flatten the lists and find unique values
unique_categories = set([category for sublist in customer_behavior['category_split'].dropna() for category in sublist])

print(unique_categories)
```

```python
customer_behavior['category_split']
```

```python
customer_behavior_categorized = customer_behavior.copy()
```

```python
#Category split

df_exploded = customer_behavior_categorized.explode('category_split')

df_exploded = df_exploded.rename(columns={'purchase_category_split': 'individual_category'})

display(df_exploded)
```

```python
#age grouped

bins = [10, 20, 30, 40, 50, 60, 70]
labels = ['10-19', '20-29', '30-39', '40-49', '50-59', '60-69']

df_exploded['age_group'] = pd.cut(df_exploded['age'], bins=bins, labels=labels, right=False)
df_exploded

df_exploded.groupby(['category_split', 'age_group'])['customer_id'].nunique()
```

```python
#improvement areas grouped

data = df_exploded['Improvement_Areas']
def categorize_area(area):
    target_groups = [
        'Customer service responsiveness',
        'Product quality and accuracy',
        'Reducing packaging waste',
        'Shipping speed and reliability'
    ]
    if area in target_groups:
        return area
    else:
        return 'Others'
df_exploded['Improvement_Areas_New'] = df_exploded['Improvement_Areas'].apply(categorize_area)
df_exploded
```

```python
#Product search method analysis by product category

cb_search_result_grouped = df_exploded.groupby(['category_split', 'Product_Search_Method']).agg({'customer_id':'nunique'})
display(cb_search_result_grouped)

fig = px.histogram (df_exploded, x='Product_Search_Method', histfunc="sum", nbins=8, text_auto=True)
fig.show()

fig2 = px.histogram(df_exploded, x = ('category_split'), barmode = 'group', color = 'Product_Search_Method', text_auto=True)
fig2.show()

fig3 = px.histogram(df_exploded, x = ('age_group'), barmode = 'group', color = 'Product_Search_Method', text_auto=True)
fig3.show()
```

```python
#Search_Result_Exploration analysis by product category

cb_product_search_grouped = df_exploded.groupby(['category_split', 'Search_Result_Exploration']).agg({'customer_id':'nunique'})
display(cb_product_search_grouped)

fig4 = px.histogram (df_exploded, x='Search_Result_Exploration', histfunc="sum", nbins=8, text_auto=True)
fig4.show()

fig5 = px.histogram(df_exploded, x = ('category_split'), histfunc="sum", barmode = 'group', color = 'Search_Result_Exploration', text_auto=True)
fig5.show()

fig6 = px.histogram(df_exploded, x = ('Gender'), histfunc="sum", barmode = 'group', color = 'Search_Result_Exploration', text_auto=True)
fig6.show()

fig7 = px.histogram(df_exploded, x = ('age_group'), histfunc="sum", barmode = 'group', color = 'Search_Result_Exploration', text_auto=True)
fig7.show()
```

```python
#Review reliability analysis by product category

cb_review_reliability = df_exploded.groupby(['Review_Reliability', 'Customer_Reviews_Importance']).agg({'customer_id':'nunique'})
display(cb_review_reliability)

fig7 = px.histogram (df_exploded, x='Review_Reliability', histfunc="sum", nbins=8, text_auto=True)
fig7.show()

fig8 = px.histogram(df_exploded, x = ('category_split'), histfunc="sum", barmode = 'group', color = 'Review_Reliability', text_auto=True)
fig8.show()

fig9 = px.histogram(df_exploded, x = ('age_group'), histfunc="sum", barmode = 'group', color = 'Review_Reliability', text_auto=True)
fig9.show()
```

```python
# relation between improvement areas and categories

fig10 = px.histogram(df_exploded, x = ('category_split'), histfunc="sum", barmode = 'group', color = 'Improvement_Areas_New', text_auto=True)
fig10.show()
```

```python
# Chi-square between improvement areas and categories

OD = pd.DataFrame()

import scipy.stats as stats

contingency_table = pd.crosstab(df_exploded['category_split'], df_exploded['Improvement_Areas_New'])

chi_res = stats.chi2_contingency(contingency_table)

print(f'p-value: {chi_res[1]}')
```

```python
cart_completion = df_exploded[['customer_id','Add_to_Cart_Browsing','Cart_Completion_Frequency','Cart_Abandonment_Factors']]
cart_completion
```

```python
# replacing words with numbers

def update(x):
  return (x).replace('Yes','1').replace('Maybe','0.5').replace('No','0')

cart_completion['Add_to_Cart_Browsing'] = cart_completion['Add_to_Cart_Browsing'].apply(update)

def update2(x):
  return (x).replace('Always','1').replace('Often','0.75').replace('Sometimes','0.5').replace('Rarely','0.25').replace('Never','0')
cart_completion['Cart_Completion_Frequency'] = cart_completion['Cart_Completion_Frequency'].apply(update2)
#cart_completion[‘Completed Purchases’] = cart_completion[‘Add_to_Cart_Browsing’] * cart_completion[‘Cart_Completion_Frequency’]
#cart_completion[‘Abandoned Purchases’] = cart_completion[‘Add_to_Cart_Browsing’] - cart_completion[‘Completed Purchases’]
cart_completion.head(30)
```

```python
# change to numeric values

cart_completion['Add_to_Cart_Browsing'] = pd.to_numeric(cart_completion['Add_to_Cart_Browsing'], errors='coerce')
cart_completion['Cart_Completion_Frequency'] = pd.to_numeric(cart_completion['Cart_Completion_Frequency'], errors='coerce')
```

```python
#labeling columns
cart_completion['completed_purchases'] = cart_completion['Add_to_Cart_Browsing'] * cart_completion['Cart_Completion_Frequency']
cart_completion['abandoned_purchases'] = cart_completion['Add_to_Cart_Browsing'] - cart_completion['completed_purchases']
cart_completion.head(10)
```

```python
df_exploded_updated = pd.merge(df_exploded, cart_completion, on ='customer_id')
```
